# **Excepciones**

Código con error. **Sin manejo de excepciones.** 

```python
lista = [0,1,2,3]

for x in range(1,6):
  print(lista[x])

print(lista)
```

Código con error. **CON manejo de excepciones.**

```python
lista = [0,1,2,3]

try:
  for x in range(1,6):
    print(lista[x])
except(IndexError):
  lista.append("NUEVO")

print(lista)
```

Ejemplo else cuando se levanta la excepción:

```python
lista = [0,1,2,3]

try:
  for x in range(1,6):
    print(lista[x])
except(IndexError):
  lista.append("NUEVO")
else:
  print("Se ejecutó el bloque try de forma correcta!")

print(lista)
```

Ejemplo else cuando **no** se levanta la excepción:

```python
lista = [0,1,2,3]

try:
  for x in range(1,4):
    print(lista[x])
except(IndexError):
  lista.append("NUEVO")
else:
  print("Se ejecutó el bloque try de forma correcta!")

print(lista)
```

Ejemplo finally **no** se levanta la excepción:

```python
def dividir(x,y):
  try:
    resultado = x / y
    print("se terminó la ejecución.")
  except(ZeroDivisionError):
    print("ERROR! División por cero.")
  else:
    print("El resultado es: ", resultado)
  finally:
    print("Ejecutando la clausula finally!!")

#invocando función
dividir(8,4)
```

Ejemplo finally, se levanta la excepción:

```python
dividir(8,0)
```

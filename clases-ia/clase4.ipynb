```python
class Mascota:

    def __init__(self, nombre, especie, edad, duenio):
        self.__nombre = nombre
        self.__especie = especie
        self.__edad = edad
        self.__duenio = duenio

    def get_nombre(self):
        return self.__nombre

    def set_nombre(self, nombre):
        self.__nombre = nombre

    def hablar_con_duenio(self):
        dicc = {'caballo': 'relinchar', 'gato': 'maullar', 'perro': 'ladrar', 'caballo': 'relinchar'}
        return dicc[self.__especie]

    def __str__(self):
        return f'Nombre: {self.__nombre}'
```

```python
mascota1 = Mascota('roco', 'perro', 2, 'pepe')
print(mascota1)
```

```python
mascota1.hablar_con_duenio()
```

```python
mascotas = [Mascota('tom','gato',3, ''), mascota1, Mascota('silver', 'caballo', 4, '')]
for mascota in mascotas:
    print(mascota, ' - ', mascota.hablar_con_duenio())
```

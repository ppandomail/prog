```python
class Punto():
  def __init__(self,x,y):
    self.x = x
    self.y = y

class FiguraGeometrica():
  def __init__(self,x,y):
    self.__centro = Punto(x,y)   ####COMPOSICION

  def __getCentro(self):
    return self.__centro

  def getX(self):
    return self.__getCentro().x

nueva_figura = FiguraGeometrica(1,5)
print(nueva_figura.getX())
```

## **Herencia Múltiple**

```python
class Terrestre:
    def desplazar(self):
      print ("El animal anda")

class Acuatico:
    def desplazar(self):
      print ("El animal nada")



class Cocodrilo(Terrestre, Acuatico):
  def __init__(self, lugar):
    Terrestre.__init__(self)
    Acuatico.__init__(self)
    self.__origen=lugar

c = Cocodrilo("Florida")
#Toma como referencia el primer elemento de la lista de parámetros
c.desplazar()
```

# **Polimorfismo**

```python
import seaborn as sb

class Grafico:
  def dibujar(self,datos,color):
    pass

class DispersionCategorica(Grafico):
  def dibujar(self,datos):
    sb.catplot(datos, x="cut", y="carat",color="green")

class Curva(Grafico):
  def dibujar(self,datos):
    sb.lineplot(datos,x="cut",y="carat",color="orange")

class Pares(Grafico):
  def dibujar(self,datos):
    sb.pairplot(datos, vars = ["cut", "carat"])
```

```python
datos = sb.load_dataset("diamonds")
print(datos)
```

```python
graficos = [Curva(),Pares(),DispersionCategorica()]

for grafico in graficos:
  grafico.dibujar(datos)
```

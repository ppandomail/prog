## 1. Manejo de Archivos

### 1.1 Apertura y Lectura de Archivos

Los archivos pueden ser procesados como archivos de texto (línea por línea) o como archivos binarios (byte por byte).

#### Abrir y leer un archivo línea por línea

```python
# Método 1: Usando readline()
archivo = open("archivo.txt")
linea = archivo.readline()
while linea != '':
    # procesar linea
    print(linea.rstrip("\n"))
    linea = archivo.readline()
archivo.close()

# Método 2: Usando un bucle for
archivo = open("archivo.txt")
for linea in archivo:
    # procesar línea
    print(linea.rstrip("\n"))
archivo.close()

# Método 3: Leer todas las líneas de una vez
archivo = open("archivo.txt")
lineas = archivo.readlines()
print(lineas)
archivo.close()
```

#### Mostrar líneas numeradas de un archivo

```python
archivo = open("archivo.txt")
i = 1
for linea in archivo:
    linea = linea.rstrip("\n")
    print("%4d: %s" % (i, linea))
    i += 1
archivo.close()
```

### 1.2 Modos de Apertura de Archivos

La función `open()` acepta diferentes modos para abrir archivos:

- `'r'`: Modo de solo lectura (predeterminado)
- `'w'`: Modo de solo escritura (vacía el archivo si existe, lo crea si no existe)
- `'a'`: Modo de agregar al final (crea el archivo si no existe)
- `'b'`: Se agrega para modo binario (ej: `'rb'`, `'wb'`)

### 1.3 Escritura de Archivos

```python
# Escribir un archivo usando write()
archivo = open("saludo.py", "w")
archivo.write('print("Hola Mundo")\n')
archivo.close()

# Escribir una lista de cadenas usando writelines()
archivo = open("nombres.txt", "w")
archivo.writelines(["Ana\n", "Juan\n"])
archivo.close()

# Agregar al final con el modo "a"
archivo = open("nombres.txt", "a")
archivo.write("María\n")
archivo.close()

# Ver el contenido del archivo
archivo = open("nombres.txt")
print(archivo.read())
archivo.close()
```

### 1.4 Sistema de Registro (Logging)

Creación de un módulo para registro de eventos:

```python
# Archivo log.py
import datetime

def abrir_log(nombre_log):
    """ Abre el archivo de log indicado. Devuelve el archivo abierto.
    Pre: el nombre corresponde a un nombre de archivo válido.
    Post: el archivo ha sido abierto posicionándose al final. """
    archivo_log = open(nombre_log, "a")
    guardar_log(archivo_log, "Iniciando registro de errores")
    return archivo_log

def guardar_log(archivo_log, mensaje):
    """ Guarda el mensaje en el archivo de log, con la hora actual.
    Pre: el archivo de log ha sido abierto correctamente.
    Post: el mensaje ha sido escrito al final del archivo. """
    hora_actual = str(datetime.datetime.now())  # Obtiene la hora actual
    archivo_log.write(hora_actual + " " + mensaje + "\n")  # Guarda mensaje

def cerrar_log(archivo_log):
    """ Cierra el archivo de log.
    Pre: el archivo de log ha sido abierto correctamente.
    Post: el archivo de log se ha cerrado. """
    guardar_log(archivo_log, "Fin del registro de errores")
    archivo_log.close()
```

```python
log = abrir_log("registro.log")
guardar_log(log, "Operación realizada con éxito")
guardar_log(log, "Se encontró un error en el proceso")
cerrar_log(log)

# Ver el contenido del registro
archivo = open("registro.log")
print(archivo.read())
archivo.close()
```

### 1.6 Persistencia de Datos

Creando un módulo para guardar y recuperar puntajes:

```python
# Archivo puntaje.py
def guardar_puntajes(nombre_archivo, puntajes):
    """ Guarda la lista de puntajes en el archivo.
    Pre: nombre_archivo corresponde a un archivo válido,
    puntajes corresponde a una lista de tuplas de 3 elementos.
    Post: se guardaron los valores en el archivo,
    separados por comas.
    """
    archivo = open(nombre_archivo, "w")
    for nombre, puntaje, tiempo in puntajes:
        archivo.write(nombre + "," + str(puntaje) + "," + tiempo + "\n")
    archivo.close()

def recuperar_puntajes(nombre_archivo):
    """ Recupera los puntajes a partir del archivo provisto.
    Devuelve una lista con los valores de los puntajes.
    Pre: el archivo contiene los puntajes en el formato esperado,
    separados por comas
    Post: la lista devuelta contiene los puntajes en el formato:
    [(nombre1,puntaje1,tiempo1),(nombre2,puntaje2,tiempo2)].
    """
    puntajes = []
    archivo = open(nombre_archivo, "r")
    for linea in archivo:
        nombre, puntaje, tiempo = linea.rstrip("\n").split(",")
        puntajes.append((nombre, int(puntaje), tiempo))
    archivo.close()
    return puntajes
```

```python
valores = [("Pepe", 108, "4:16"), ("Juana", 2315, "8:42")]
guardar_puntajes("puntajes.txt", valores)

# Contenido del archivo
archivo = open("puntajes.txt")
print(archivo.read())
archivo.close()

recuperado = recuperar_puntajes("puntajes.txt")
print(recuperado)  # [('Pepe', 108, '4:16'), ('Juana', 2315, '8:42')]
```

Si un nombre contiene una coma, la línea ya no tiene tres valores y no se puede recuperar:

```python
guardar_puntajes("puntajes.txt", [("Pérez, Ana", 450, "5:03")])

try:
    print(recuperar_puntajes("puntajes.txt"))
except ValueError:
    print("Error: La línea no tiene tres valores separados por comas.")
```

### 1.7 Archivos CSV

Los archivos CSV (Comma Separated Values) son un formato común para intercambiar datos. El módulo `csv` de Python separa los campos y respeta los valores entre comillas:

```python
import csv

# Escribir un archivo CSV
def escribir_csv(nombre_archivo, datos):
    archivo = open(nombre_archivo, "w")
    # Escribir encabezados
    archivo.write("Nombre,Apellido,Telefono,Cumpleaños\n")
    # Escribir datos
    for nombre, apellido, telefono, cumpleaños in datos:
        archivo.write('"' + nombre + '","' + apellido + '","' + telefono + '","' + cumpleaños + '"\n')
    archivo.close()

# Leer un archivo CSV
def leer_csv(nombre_archivo):
    filas = []
    archivo = open(nombre_archivo, newline="")
    for fila in csv.reader(archivo):
        filas.append(fila)
    archivo.close()
    return filas

# Ejemplo de uso
personas = [
    ("John", "Smith", "555-0101", "1973-11-24"),
    ("Jane", "Smith", "555-0101", "1975-06-12")
]
escribir_csv("contactos.csv", personas)
filas = leer_csv("contactos.csv")
print("Encabezados:", filas[0])
print("Datos:", filas[1:])
```

También se puede escribir con `csv.writer`, que agrega comillas solo cuando un valor las necesita:

```python
archivo = open("puntajes.csv", "w", newline="")
escritor = csv.writer(archivo)
escritor.writerows([("Pepe", 108, "4:16"), ("Pérez, Ana", 450, "5:03")])
archivo.close()

# Contenido del archivo
archivo = open("puntajes.csv")
print(archivo.read())
archivo.close()

print(leer_csv("puntajes.csv"))
```

## 2. Matrices en Python

Una matriz es una estructura de datos bidimensional que consta de filas y columnas.

### 2.1 Representación de Matrices

```python
# Definir una matriz usando listas anidadas
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Acceder a elementos de la matriz
print("Elemento en la posición [0][1]:", matriz[0][1])  # Salida: 2
print("Fila de índice 1:", matriz[1])  # Salida: [4, 5, 6]

# Imprimir toda la matriz de forma estructurada
print("Matriz completa:")
for fila in matriz:
    print(fila)
```

### 2.2 Operaciones Básicas con Matrices

#### Suma de Matrices

```python
def suma_matrices(matriz1, matriz2):
    resultado = []
    for i in range(len(matriz1)):
        fila = []
        for j in range(len(matriz1[0])):
            fila.append(matriz1[i][j] + matriz2[i][j])
        resultado.append(fila)
    return resultado

# Ejemplo de uso
matriz1 = [[1, 2], [3, 4]]
matriz2 = [[5, 6], [7, 8]]
resultado = suma_matrices(matriz1, matriz2)
print("Matriz 1:", matriz1)
print("Matriz 2:", matriz2)
print("Suma:", resultado)  # Salida: [[6, 8], [10, 12]]
```

#### Multiplicación Escalar

```python
def multiplicacion_escalar(matriz, escalar):
    resultado = []
    for i in range(len(matriz)):
        fila = []
        for j in range(len(matriz[0])):
            fila.append(matriz[i][j] * escalar)
        resultado.append(fila)
    return resultado

# Ejemplo de uso
matriz = [[1, 2], [3, 4]]
escalar = 2
resultado = multiplicacion_escalar(matriz, escalar)
print("Matriz original:", matriz)
print("Escalar:", escalar)
print("Resultado:", resultado)  # Salida: [[2, 4], [6, 8]]
```

### 2.3 Operaciones con NumPy

NumPy es una biblioteca para procesamiento numérico en Python. En Colab ya viene instalada. En otro entorno se instala con `pip install numpy`.

```python
import numpy as np

# Crear matrices con NumPy
matriz1 = np.array([[1, 2], [3, 4]])
matriz2 = np.array([[5, 6], [7, 8]])

print("Matriz 1:\n", matriz1)
print("Matriz 2:\n", matriz2)

# Producto Punto (Dot Product)
resultado_dot = np.dot(matriz1, matriz2)
print("Producto punto:\n", resultado_dot)

# Multiplicación de matrices
resultado_matmul = np.matmul(matriz1, matriz2)
print("Multiplicación de matrices:\n", resultado_matmul)

# Transposición de matrices
matriz_transpuesta = matriz1.T
print("Matriz transpuesta:\n", matriz_transpuesta)

# Inversa de una matriz
matriz_inversa = np.linalg.inv(matriz1)
print("Matriz inversa:\n", matriz_inversa)
```

## 3. Otras Estructuras de Datos

### 3.1 Grafos

Un grafo es una estructura de datos que consta de nodos (vértices) y aristas que los conectan.

#### Implementación Básica con Diccionarios

```python
# Implementación de un grafo usando diccionarios
grafo = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

# Recorrer los nodos del grafo
print("Nodos del grafo:", list(grafo.keys()))

# Ver los vecinos de un nodo
print("Vecinos del nodo A:", grafo['A'])
print("Vecinos del nodo B:", grafo['B'])

# Verificar si existe una conexión entre dos nodos
def existe_conexion(grafo, nodo1, nodo2):
    return nodo2 in grafo[nodo1]

print("¿Existe conexión entre A y B?", existe_conexion(grafo, 'A', 'B'))
print("¿Existe conexión entre A y D?", existe_conexion(grafo, 'A', 'D'))
```

### 3.2 Árboles

Un árbol es una estructura jerárquica donde un nodo raíz se conecta a varios nodos hijos.

#### Implementación Básica

```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.hijos = []

    def agregar_hijo(self, hijo):
        self.hijos.append(hijo)

    def __str__(self, nivel=0):
        resultado = "  " * nivel + str(self.valor) + "\n"
        for hijo in self.hijos:
            resultado += hijo.__str__(nivel + 1)
        return resultado

# Crear nodos
raiz = Nodo('A')
nodo_b = Nodo('B')
nodo_c = Nodo('C')
nodo_d = Nodo('D')

# Construir el árbol
raiz.agregar_hijo(nodo_b)
raiz.agregar_hijo(nodo_c)
nodo_b.agregar_hijo(nodo_d)

# Imprimir el árbol
print(raiz)
```

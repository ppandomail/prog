```python
numeros = []
suma = 0
for _ in range(10):
    numero = int(input('Ingrese número: '))
    suma += numero
    numeros.append(numero)
promedio = suma/len(numeros)
for elem in numeros:
    if elem > promedio:
        print(elem)
```

```python
alumnos = {}

respuesta = input('¿Ingresar alumno? si/no: ')
while respuesta == 'si':
    dni = input('DNI: ')
    promedio = float(input('Promedio'))
    alumnos[dni] = promedio
    respuesta = input('¿Ingresar alumno? si/no: ')

if len(alumnos) == 0:
    print('No hay alumnos')
else:
    promedios = list(alumnos.values())
    promedios.sort()
    mejor_promedio = promedios[-1]
    print('Mejor promedio: ', mejor_promedio)

    for dni, prom in alumnos.items():
        if prom == mejor_promedio:
            print('DNI', dni)    
```

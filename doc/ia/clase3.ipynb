```python
def promedio(lista):
    suma = 0
    for precio in lista:
        suma += precio
    return suma/len(lista)
```

```python
promedio([200, 450, 180])
```

```python
def fact(n):
    if n==0:
        return 1
    return n * fact(n-1)
```

```python
fact(3)
```

```python
fact(4)
```

```python
def suma(n):
    if n==1:
        return n
    return n + suma(n-1)
```

```python
suma(10)
```

```python
def imprimir(lista, pos):
    if pos != -1:
        print(lista[pos])
        imprimir(lista, pos-1)
```

```python
lista = ['a', 'b']
imprimir(lista, len(lista)-1)
```

```python
def imprimir2(lista, pos=0):
    if pos < len(lista):
        print(lista[pos])
        imprimir2(lista, pos+1)
```

```python
imprimir2(lista)
```

```python
def imprimir_lista(lista, posicion):
  if(posicion != -1):
    print(lista[posicion])
    imprimir_lista(lista,posicion - 1)

lista = ["perez","sand","arlt","claymore"]
imprimir_lista(lista,len(lista) - 1)
```

```python
def imprimir_lista(lista, posicion):
  if(posicion < len(lista)):
    print(lista[posicion])
    imprimir_lista(lista,posicion + 1)

lista = ["perez","sand","arlt","claymore"]
imprimir_lista(lista,0)
```

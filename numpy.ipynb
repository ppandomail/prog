```python
# altura en metros de personas
alturas = [1.73, 1.68, 1.71, 1.89, 1.79, 1.76, 1.67]

# peso en Kg de personas
pesos = [65.4, 59.2, 63.6, 88.4, 68.7, 89.7, 73.2]

# imc = pesos_personas_en_kg / (alturas_personas_en_metros ** 2)
```

## Sin NumPy

```python
persona = 0
for altura in alturas:
    peso = pesos[persona]
    imc = round(peso / (altura ** 2), 8)
    print(imc)
    persona = persona + 1
```

## Estadística básica

[Documentación de ndarray](https://numpy.org/doc/2.2/reference/generated/numpy.ndarray.html#numpy.ndarray)

```python
valores = np.array([100, 200, 200, 250, 600])
```

```python
# suma
valores.sum()
```

```python
# media o promedio
valores.mean()
```

```python
# mínimo
valores.min()
```

```python
# máximo
valores.max()
```

```python
# mediana: valor del punto medio
np.median(valores)
```

```python
# desvio: cuanto se desvian los valores de la media en la distribución
# describe que tan dispersos están los valores
# si es baja -> números cerca del valor medio
# si es alta -> valores distribuidos en un rango más amplio
valores.std()
```

```python
# varianza: indica que tan dispersos están los valores
valores.var()
```

```python
# percentiles: se usan para proporcionar un número que describe el valor al cual un 
# porcentaje determinado de los valores es inferior
ages = [5, 31, 43, 48, 50, 41, 7, 11, 15, 39, 80, 82, 32, 2, 8, 6, 25, 36, 27, 61, 31]

percentil75 = np.percentile(ages, 75)
percentil75 # La respuesta es 43, lo que significa que el 75 % de las personas tienen 43 años o menos
```

```python
from scipy import stats

# moda: valor más común
moda = stats.mode(ages)
moda
```

## Coeficientes de correlación

```python
# calcula el coeficiente de correlación de Pearson
# mide la correlación lineal entre dos variables x, y
# 1: si los datos están correlacionados (si x aumenta, entonces aumenta y)
np.corrcoef(numpy_2d[:, 0], numpy_2d[:, 1])
```

## Ordenar

```python
# ordenar un np array
array_fibonacci = np.array([55, 0, 144, 1, 21, 89, 5, 8, 13, 1, 34, 3, 2])
```

```python
# orden ascendente
np.sort(array_fibonacci)
```

```python
# orden descendente
-np.sort(-array_fibonacci)
```

```python
# ordenar un 2D np array sobre el primer eje (columnas)
a = np.array([[10, 40, 30, 20],[30, 20, 10, 40]])
np.sort(a, axis=0)
```

```python
# ordenar un 2D np array sobre el último eje (filas)
np.sort(a, axis=1)
```

```python
# aplanar y ordenar
np.sort(a, axis=None)
```

## Arrays con distintos tipos de datos

```python
tipos = [('nom', 'S15'), ('ape', 'S15'), ('com', int), ('nota', float)]
alumnos = [('pablo', 'pandolfo', 1, 10.0),
           ('maria', 'garcia', 2, 8.0), 
           ('ana', 'perez', 2, 5.0)]

alumnos = np.array(alumnos, dtype=tipos)
alumnos
```

```python
np.sort(alumnos, order='ape')
```

```python
np.sort(alumnos, order='nota')
```

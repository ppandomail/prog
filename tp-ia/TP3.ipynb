**Defina una función que reciba una palabra y retorne la cantidad de vocales que contiene**
* **a. Implemente la solución con una función iterativa**
* **b. Implemente la solución con una función recursiva**

```python
def cant_vocales_iterativa(palabra):
    cant = 0
    for simbolo in palabra:
        if simbolo in ['a', 'á', 'e', 'é', 'i', 'í', 'o', 'ó', 'u', 'ú']:
            cant += 1
    return cant
```

```python
cant_vocales_iterativa('pepe')
```

```python
cant_vocales_iterativa('inteligencia')
```

```python
def cant_vocales_recursiva(palabra, posicion=0, cant=0):
    if posicion < len(palabra):
        if palabra[posicion] in ['a', 'á', 'e', 'é', 'i', 'í', 'o', 'ó', 'u', 'ú']:
            cant += 1
        return cant_vocales_recursiva(palabra, posicion + 1, cant)
    return cant    
```

```python
cant_vocales_recursiva('pepe')
```

```python
cant_vocales_recursiva('inteligencia')
```

**4. Implementar una función que reciba como parámetro un número entero y retorne la cantidad de dígitos que posee y la suma de los mismos**

```python
def cant_digitos_suma(entero):
    lista = []
    for digito in str(entero):
        lista.append(int(digito))
    return len(lista), sum(lista)
```

```python
cant_digitos_suma(1234)
```

**5. Escribir un programa que genere aleatoriamente 100 números reales, los almacene en una lista, y luego los muestre ordenados de menor a mayor. Obligatorio utilizar funciones**

```python
import random

def aleatorios_ordenados():
    lista = []
    for _ in range(0, 100):
        lista.append(random.random())
    return sorted(lista)
```

**6. Escriba una función recursiva que reciba una lista de apellidos y retorne si existe o no un apellido dado en la lista**

```python
def busqueda_secuencial_recursiva(lista, objetivo, indice=0):
    if indice >= len(lista):
        return False
    if lista[indice] == objetivo:
        return True
    return busqueda_secuencial_recursiva(lista, objetivo, indice + 1)
```

```python
busqueda_secuencial_recursiva(['perez', 'garcia', 'fernandez', 'diaz'], 'pepe')
```

```python
busqueda_secuencial_recursiva(['perez', 'garcia', 'fernandez', 'diaz'], 'garcia')
```

**7. Escriba una función recursiva que reciba un número y una lista de números y retorne la cantidad de veces que dicho número aparece en la lista**

```python
def cantidad_recursiva(numero, lista, indice=0, cant=0):
    if indice >= len(lista):
        return cant
    if lista[indice] == numero:
        cant += 1
    return cantidad_recursiva(numero, lista, indice + 1, cant)
```

```python
cantidad_recursiva(4, [1, 2, 3, 4, 5])
```

```python
cantidad_recursiva(3, [1, 2, 3, 3, 5])
```

**8. Realizar un programa que reciba como parámetro el radio de un círculo y retorne su diámetro y su perímetro. Implementar utilizando funciones**

```python
import math

def diametro(radio):
    return 2 * radio

def perimetro(radio):
    return math.pi * diametro(radio)

def diametro_perimetro(radio):
    return diametro(radio), perimetro(radio)
```

```python
diametro_perimetro(2)
```

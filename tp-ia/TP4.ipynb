**1. Escriba en Python una clase llamada Temperatura que tenga un método denominado calcularCelsius, y que dada una temperatura en grados
Fahrenheit devuelva la misma en Celsius y otro método llamado calcularFahrenheit que haga el cálculo inverso. Luego escriba un programa que ingrese por teclado dos temperaturas, una en Fahrenheit y otra en Celsius, e imprima la conversión a Celsius y a Fahrenheit respectivamente**

```python
class Temperatura:

    def calcular_celsius(self, valor):
        return (valor - 32) * 5/9

    def calcular_fahrenheit(self, valor):
        return (valor + 9/5) + 32

fahrenheit = float(input('Temperatura en Fahrenheit: '))
celsius = float(input('Temperatura en Celsius: '))
temperatura = Temperatura()
print(temperatura.calcular_celsius(fahrenheit))
print(temperatura.calcular_fahrenheit(celsius))
```

**2. Defina una clase Persona con 3 atributos: Nombre, Edad y DNI. Escriba un algoritmo que permita al usuario ingresar en una consola los datos de una serie de personas y guardarlas en una lista. El proceso de entrada finaliza con un
string vacío. Una vez finalizada la entrada de datos, el programa debe imprimir en la consola un listado con la forma (Pista: usar str ):
Ejemplo:** 
* **Juan Perez (40) 2098745** 
* **José García (41) 1965412**

```python
class Persona:

    def __init__(self, nombre, edad, dni):
        self.__nombre = nombre
        self.__edad = edad
        self.__dni = dni

    def __str__(self):
        return f'{self.__nombre} ({self.__edad}) {self.__dni}'

personas = []
nombre = input('Nombre: ')
while nombre != '':
    edad = int(input('Edad: '))
    dni = input('DNI: ')
    personas.append(Persona(nombre, edad, dni))
    nombre = input('Nombre: ')
for persona in personas:
    print(persona)
```

**3. Escriba en Python un módulo llamado calendario donde esté codificada la clase Fecha de acuerdo con las siguientes especificaciones:**
* **a) Atributos: día, mes, año**
* **b) Funcionalidad: se debe permitir modificar la fecha, se debe poder mostrar la fecha en formato numérico (día/mes/año)**
* **c) Al crear una fecha se inicializa con valores para el día, el mes y el año**
* **d) Identificar el estado y el comportamiento de los objetos de tipo Fecha**
**En una aplicación crear dos fechas y calcular e imprimir la diferencia en días entre ambas fechas. Ayuda: pasar ambas fechas a días (año x 365 + mes x 30 + dia)**

```python
class Fecha:

    def __init__(self, dia, mes, anio):
        self.__dia = dia
        self.__mes = mes
        self.__anio = anio

    def set_dia(self, dia):
        self.__dia = dia

    def set_mes(self, mes):
        self.__mes = mes

    def set_anio(self, anio):
        self.__anio = anio

    def calcular_diferencia_en_dias(self, fecha):
        dias1 = self.__anio * 365 + self.__mes * 30 + self.__dia
        dias2 = fecha.__anio * 365 + fecha.__mes * 30 + fecha.__dia
        return abs(dias1 - dias2)        
    
    def __str__(self):
        return f'{self.__dia}/{self.__mes}/{self.__anio}'

fecha1 = Fecha(9,11,1975)
fecha2 = Fecha(18,9,2026)
print(fecha1)
print(fecha1.calcular_diferencia_en_dias(fecha2))
```

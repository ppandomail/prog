**1. Escriba un programa que solicite al estudiante tres calificaciones (números del 0 al 10), calcule el promedio y muestre si el estudiante ha aprobado (promedio >= 6) o no**

```python
nota1 = int(input('Nota 1: '))
nota2 = int(input('Nota 2: '))
nota3 = int(input('Nota 3: '))
promedio = (nota1 + nota2 + nota3) / 3
if promedio >= 6:
    print('Aprobado')
else:
    print('Desaprobado')
```

```python
def calificacion(nota1, nota2, nota3):
    return 'Aprobado' if ((nota1 + nota2 + nota3) / 3) >= 6 else 'Desaprobado'
calificacion(4, 5, 6)
```

**2. Crea un programa que permita convertir una temperatura de grados Celsius a Fahrenheit o viceversa, según lo elija el usuario**

```python
temperatura = float(input('Temperatura: '))
opcion = int(input('1. ºC -> ºF | 2. ºF -> ºC'))
if opcion == 1:
    temperatura = (temperatura + 9/5) + 32
else:
    temperatura = (temperatura - 32) * 5/9
print(temperatura)
```

```python
def c_f(temp):
    return (temp + 9/5) + 32

def f_c(temp):
    return (temp - 32) * 5/9

temperatura = float(input('Temperatura: '))
opcion = int(input('1. ºC -> ºF | 2. ºF -> ºC'))
c_f(temperatura) if opcion == 1 else f_c(temperatura)
```

**3. Desarrolla un programa que solicite un número entero al usuario y determine si es positivo, negativo o cero, y además indique si es par o impar (exceptuando el cero)**

```python
numero = int(input('Número: '))
if numero == 0:
    print('cero')
elif numero > 0:
    print('positivo')
else:
    print('negativo')
if numero != 0:
    resto = numero % 2
    if resto == 0:
        print('par')
    else:
        print('impar')
```

```python
def pos_neg_cero(numero):
    return 'cero' if numero == 0 else 'positivo' if numero > 0 else 'negativo'

def par_impar(numero):
    return None if numero == 0 else 'par' if numero%2 == 0 else 'impar'

def eval(numero):
    return pos_neg_cero(numero), par_impar(numero)

eval(-8)
```

**4. Escribe un programa para una tienda que calcule el precio final de un producto después de aplicar descuentos según el monto de compra:**
* **Si el monto es menor a 1000: no hay descuento**
* **Si el monto está entre 1000 y 5000: 10% de descuento**
* **Si el monto es mayor a 5000: 20% de descuento**

```python
def precio_final(monto):
    if monto < 1000:
        return monto
    elif monto >= 1000 and monto <= 5000:
        return monto * 0.9
    return monto * 0.8

precio_final(2000)
```

**5. Escriba un programa que pida un año por pantalla y diga si es bisiesto. Un año es bisiesto si es divisible por 4 pero no por 100. Si es divisible por 100, para ser bisiesto debe ser divisible por 400**

```python
def is_bisiesto(anio):
    if anio % 4 == 0 and anio % 100 != 0:
        return True
    if anio % 100 == 0 and anio % 400 == 0:
        return True
    return False

anio = int(input('Año: '))
is_bisiesto(anio)
```

```python
def is_bisiesto_v2(anio):
    return (anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0)

anio = int(input('Año: '))
is_bisiesto_v2(anio)
```

**1. Derive de la clase Persona definida en la práctica 5 la clase Alumno, agréguele un atributo privado llamado promedio. Defina los constructores necesarios. Luego, modifique el programa del ejercicio 3 de la práctica anterior para tratar ahora con
alumnos. El listado debe tener la siguiente apariencia:**
* **1) ALUMNO: Juan Perez (26) 2998745 7.8**
* **2) ALUMNO: Jose García (25) 3065412 8.4**

```python
class Persona:

    def __init__(self, nombre, edad, dni):
        self.__nombre = nombre
        self.__edad = edad
        self.__dni = dni

    def __str__(self):
        return f'{self.__nombre} ({self.__edad}) {self.__dni}'
```

```python
class Alumno(Persona):

    def __init__(self, nombre, edad, dni, promedio):
        super().__init__(nombre, edad, dni)
        self.__promedio = promedio

    def __str__(self):
        return f'{super().__str__()} {self.__promedio}'
```

```python
print('ALUMNO', Alumno('Juan Perez', 26, 2998745, 7.8))
print('ALUMNO', Alumno('Jose Garcia', 25, 3065412, 8.4))
```

**3. Cree la siguiente jerarquía de clases: de OperacionAritmetica. Implemente un método en cada subclase que reciba dos números y devuelva el resultado correspondiente**

```python
class OperacionAritmetica:

    def operar(self, op1, op2):
        pass

class Suma(OperacionAritmetica):

    def operar(self, op1, op2):
        return op1 + op2

class Resta(OperacionAritmetica):

    def operar(self, op1, op2):
        return op1 - op2

class Multiplicacion(OperacionAritmetica):

    def operar(self, op1, op2):
        return op1 * op2

class Division(OperacionAritmetica):

    def operar(self, op1, op2):
        return op1 / op2
```

```python
lista = [Suma(), Suma(), Resta(), Division(), Multiplicacion()]
for op in lista:
    print(op.operar(4, 2))
```

**3) Queremos representar la información de empleados de un club: jugadores y entrenadores**
* **Cualquier empleado se caracteriza por su nombre y sueldo básico**
* **Los jugadores son empleados que se caracterizan por el número de partidos jugados y el número de goles anotados**
* **Los entrenadores son empleados que se caracterizan por la cantidad de campeonatos ganados**
* **Para los jugadores: el sueldo a cobrar es el sueldo básico y si el promedio de goles por partido es superior a 0,5 se adiciona un plus de otro sueldo básico**
* **Para los entrenadores: el sueldo a cobrar es el sueldo básico al cual se le adiciona un plus por campeonatos ganados (5000 si ha ganado entre 1 y 4 campeonatos; 30.000 si ha ganado entre 5 y 10 campeonatos; 50.000 si ha ganado más de 10 campeonatos)**
    * **A. Implemente la jerarquía de clases, con los atributos de cada clase y métodos para obtener/modificar el valor de estos**
    * **B. Implemente constructores para los jugadores y entrenadores, que reciban toda la información necesaria para inicializar el objeto en cuestión**
    * **C. Cualquier empleado (jugador / entrenador) debe saber responder al mensaje calcularSueldoACobrar (que calcula y devuelve el sueldo a cobrar) pero de manera diferente:**

```python
class Empleado:

    def __init__(self, nombre, sueldo_basico):
        self.__nombre = nombre
        self.__sueldo_basico = sueldo_basico

    def get_nombre(self):
        return self.__nombre

    def set_nombre(self, nombre):
        self.__nombre = nombre

    def get_sueldo_basico(self):
        return self.__sueldo_basico

    def set_sueldo_basico(self, sueldo_basico):
        self.__sueldo_basico = sueldo_basico
```

```python
class Jugador(Empleado):

    def __init__(self, nombre, sueldo_basico, num_partidos_jugados, num_goles_anotados):
        super().__init__(nombre, sueldo_basico)
        self.__num_partidos_jugados = num_partidos_jugados
        self.__num_goles_anotados = num_goles_anotados

    def get_num_partidos_jugados(self):
        return self.__num_partidos_jugados

    def set_num_partidos_jugados(self, num_partidos_jugados):
        self.__num_partidos_jugados = num_partidos_jugados

    def get_num_goles_anotados(self):
        return self.__num_goles_anotados

    def set_num_goles_anotados(self, num_goles_anotados):
        self.__num_goles_anotados = num_goles_anotados

    def promedio_goles(self):
        return self.__num_goles_anotados / self.__num_partidos_jugados

    def calcular_sueldo_a_cobrar(self):
        if self.promedio_goles() > 0.5:
            return self.get_sueldo_basico() * 2
        return self.get_sueldo_basico()
```

```python
class Entrenador(Empleado):

    def __init__(self, nombre, sueldo_basico, cant_campeonatos_ganados):
        super().__init__(nombre, sueldo_basico)
        self.__cant_campeonatos_ganados = cant_campeonatos_ganados

    def get_cant_campeonatos_ganados(self):
        return self.__cant_campeonatos_ganados

    def set_cant_campeonatos_ganados(self, cant_campeonatos_ganados):
        self.__cant_campeonatos_ganados = cant_campeonatos_ganados

    def plus(self):
        if self.__cant_campeonatos_ganados >= 1 and self.__cant_campeonatos_ganados <= 4:
            return 5000
        if self.__cant_campeonatos_ganados >= 5 and self.__cant_campeonatos_ganados <= 10:
            return 30000
        if self.__cant_campeonatos_ganados >= 11:
            return 50000
        return 0

    def calcular_sueldo_a_cobrar(self):
        return self.get_sueldo_basico() + self.plus()
```
